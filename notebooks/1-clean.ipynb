## Clean Film Festivals Data

**👉 Task: Remove columns you don\'t need**


```python
festivals = pd.read_csv('../data/raw/1_film-dataset_festival-program_wide.csv')
festivals.head()
```

```python
# Keep only the columns we need
festivals = festivals[['imdb.id', 'title.mixed', 'sample.year.first', 'fest.first']]

# Drop rows missing imdb.id
festivals = festivals.dropna(subset=['imdb.id'])
print(f"Rows remaining: {len(festivals)}")
festivals.head()
```

```python
# Read addendum and keep matching columns
addendum = pd.read_csv('../data/raw/film-festivals-addendum.csv')
addendum = addendum[['imdb.id', 'title.mixed', 'sample.year.first', 'fest.first']]
addendum = addendum.dropna(subset=['imdb.id'])

# Merge with festivals
festivals = pd.concat([festivals, addendum], ignore_index=True)
festivals = festivals.drop_duplicates(subset=['imdb.id'])
print(f"Total rows after merge: {len(festivals)}")
festivals.head()
```

```python
# Standardize fest.first names
fest_mapping = {
    'BERLINALE': 'BERLINALE',
    'CANNES': 'CANNES',
    'CLERMONT-FERRAND': 'CLERMONT-FERRAND',
    'FRAMELINE': 'FRAMELINE',
    'IDFA': 'IDFA',
    'TIFF': 'TIFF',
}

# Uppercase for matching, then map to standard names
festivals['fest.first'] = festivals['fest.first'].str.upper().map(fest_mapping)

# Drop rows that don't match any of the 6 festivals
festivals = festivals.dropna(subset=['fest.first'])
print(f"Rows after standardizing: {len(festivals)}")
print(festivals['fest.first'].value_counts())
```

```python
# Export cleaned festivals data
festivals.to_csv('../data/processed/festivals_cleaned.csv', index=False)
```

# Merge Data

**👉 Task: Merge two datasets so you only have the rows and columns you need**

```python
# Match festivals with oscars by IMDB ID
# Every film in oscars_cleaned is already nominated
oscar_ids = set(oscars['FilmId'].dropna().str.split('|').explode().str.strip())

# Add oscar_nominated column to festivals
festivals['oscar_nominated'] = festivals['imdb.id'].isin(oscar_ids).astype(int)

festivals.head()
```

```python
# Calculate % of Oscar-nominated films per festival per year
nomination_pct = (
    festivals
    .groupby(['fest.first', 'sample.year.first'])['oscar_nominated']
    .apply(lambda x: x.sum() / len(x) * 100)
    .reset_index()
    .rename(columns={'oscar_nominated': 'oscar_nominated_pct'})
)

nomination_pct.head(10)
```

```python
# Export nomination percentage by festival and year
nomination_pct.to_csv('../data/processed/nomination_pct_by_festival_year.csv', index=False)
print("Exported: nomination_pct_by_festival_year.csv")
```

```python
# Pivot: one column per festival, rows are years
pivot = nomination_pct.pivot(index='sample.year.first', columns='fest.first', values='oscar_nominated_pct')

# Add "Any festival" column (overall % across all festivals per year)
any_fest = (
    festivals
    .groupby('sample.year.first')['oscar_nominated']
    .apply(lambda x: x.sum() / len(x) * 100)
)
pivot['Any festival'] = any_fest

# Clean up
pivot = pivot.reset_index().rename(columns={'sample.year.first': 'Year_int'})
pivot = pivot[['Year_int', 'Any festival', 'BERLINALE', 'CANNES', 'CLERMONT-FERRAND', 'FRAMELINE', 'IDFA', 'TIFF']]

pivot.to_csv('../data/processed/nomination_pct_wide.csv', index=False)
print("Exported: nomination_pct_wide.csv")
pivot
```

```python
# Export merged data
merged = festivals[['imdb.id', 'title.mixed', 'sample.year.first', 'fest.first', 'oscar_nominated']]
merged.to_csv('../data/processed/merged_festivals_oscars.csv', index=False)
print("Exported: merged_festivals_oscars.csv")
```
